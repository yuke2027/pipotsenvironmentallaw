```python
import pandas as pd
import numpy as np

def load_features_csv(file_path):
    """
    读取每行只有一列，内容为'数字,数字,数字'字符串的CSV文件，返回二维numpy数组
    """
    df = pd.read_csv(file_path, header=None)
    # 跳过表头（第一行），只处理数据行
    data_rows = df.iloc[1:, 0]
    # 拆分每一行字符串为数字
    X = np.array([list(map(float, row.split(','))) for row in data_rows], dtype=np.float32)
    return X

# 文件路径
train_path = r"E:\24-25-2\环境与资源保护法\基于pypots时间序列框架的空气质量数据分析张瀚元\处理后数据\train_imputed.csv"
val_path = r"E:\24-25-2\环境与资源保护法\基于pypots时间序列框架的空气质量数据分析张瀚元\处理后数据\val_imputed.csv"
test_path = r"E:\24-25-2\环境与资源保护法\基于pypots时间序列框架的空气质量数据分析张瀚元\处理后数据\test_imputed.csv"

# 加载数据
train_X = load_features_csv(train_path)
val_X = load_features_csv(val_path)
test_X = load_features_csv(test_path)

print("train_X shape:", train_X.shape)
print("val_X shape:", val_X.shape)
print("test_X shape:", test_X.shape)
```

```python
import pandas as pd
import numpy as np

def load_X_y_from_csv(file_path):
    df = pd.read_csv(file_path, header=None)
    data_rows = df.iloc[1:, 0]
    data = np.array([list(map(float, row.split(','))) for row in data_rows], dtype=np.float32)
    X = data[:, :-1]  # 除最后一列外都是特征
    y = data[:, -1].astype(np.int64)  # 最后一列是标签
    return X, y

# 文件路径
train_path = r"E:\24-25-2\环境与资源保护法\基于pypots时间序列框架的空气质量数据分析张瀚元\处理后数据\train_imputed.csv"
val_path = r"E:\24-25-2\环境与资源保护法\基于pypots时间序列框架的空气质量数据分析张瀚元\处理后数据\val_imputed.csv"
test_path = r"E:\24-25-2\环境与资源保护法\基于pypots时间序列框架的空气质量数据分析张瀚元\处理后数据\test_imputed.csv"

train_X, train_y = load_X_y_from_csv(train_path)
val_X, val_y = load_X_y_from_csv(val_path)
test_X, test_y = load_X_y_from_csv(test_path)

print("train_X shape:", train_X.shape)
print("train_y shape:", train_y.shape)
print("val_X shape:", val_X.shape)
print("val_y shape:", val_y.shape)
print("test_X shape:", test_X.shape)
print("test_y shape:", test_y.shape)
```
